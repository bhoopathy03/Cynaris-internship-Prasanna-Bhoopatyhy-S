```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
import statsmodels.api as sm
from statsmodels.formula.api import ols
```

```python
df=pd.read_csv("/content/ecommerce_customer_behavior_dataset.csv")
df.head(2)
```

# 1. customer satsifaction hypothesis
  ho: the average customer satsfication is 4
  
  h1: The average customer satsfication is less than 4

```python
# lets take the customer satsfication column for hypothesis
# now we need to find mean of the column
Mean_customer_satsfication=df['Customer_Rating'].mean()

# perform one sample t-test
t_stat,p_value=stats.ttest_1samp(df['Customer_Rating'],4)

print("T-statistic:",t_stat,"\nP-value:",p_value)
```

```python
alpha =0.05

if p_value<alpha:
	print("Reject the null hypothesis","\nThe average customer satsfication is 4")
else:
	print("Failt to reject the Null hypothesis","\nThe average customer satsfication is less than 4")
```

# confidence interval for one sample

```python
mean= df['Customer_Rating'].mean()
ci=stats.t.interval(confidence=0.95,df=len(df['Customer_Rating'])-1,loc=mean,scale=stats.sem(df['Customer_Rating']))
print("confidence interval:",ci)
```

# 2. For two sample test let's test the total sales of men and women are equal
  H0: The sales of men and women are equal
  
  H1: The sales of men and women are not equal

```python
male_sales = df[df['Gender'] == 'Male']['Total_Amount']
female_sales = df[df['Gender'] == 'Female']['Total_Amount']
```

```python
t_stat,p_value= stats.ttest_ind(
	male_sales,
	female_sales,
	equal_var=False)

print("T-statistic:",t_stat,"\nP-value:",P_value)
```

```python
alpha =0.05

if p_value<alpha:
	print("Reject the null hypothesis","\nThe sales of men and women are equal")
else:
	print("Failt to reject the Null hypothesis","\nThe sales of men and women are not equal")
```

# CI for two samples

```python
Male = sm.stats.DescrStatsW(male_sales)
Female = sm.stats.DescrStatsW(female_sales)

cm = sm.stats.CompareMeans(Male, Female)
ci = cm.tconfint_diff(usevar='unequal')
print("Confidence Interval:", ci)
```

## Type 1 error: Rejecting Null hypothesis when its actual true

## Type 2 error: Fail to reject when hypothesis when it's actual not true
