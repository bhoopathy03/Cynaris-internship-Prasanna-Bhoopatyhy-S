```python
# import pandas as matplotlib
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Create 2D array of table given:
Data=[['1','M',34,123,'Normal',350],['2','F',40,114,'Overweight',450],
      ['3','F',34,135,'Obesity',169],[ '4','M',30,139,'Underweight',189],
      ['5','F',44,117,'Underweight',183],['6','M',36,121,'Normal',80],
      ['7','M',32,133,'Obesity',166], ['8','F',26,140,'Normal',120],
      ['9','M',32,133,'Normal',75],[ '10','M',36,133,'Underweight',40]]

#dataframe created with the above data array
df=pd.DataFrame(Data,columns=['Sl.No','Gender','Age','Sales','BMI','Income'])
```

```python
df.head()
```

## creating line plot using age and sales

```python
# Create a line plot
plt.plot(df['Age'], df['Sales'], marker='o')

plt.xlabel('Age')
plt.ylabel('Sales')
plt.title('Age vs Sales')
plt.grid(True)

plt.show()
```

## Bar plot for the attributes using matplotlib And saving the graph in png

```python
df.plot.bar()
plt.bar(df['Age'],df['Sales'])
plt.xlabel("Age")
plt.ylabel("Sales")
plt.show()
plt.savefig("sample_1.png")
```

## Scatter chart using matplotlib

```python
plt.scatter(df['Income'],df['Age'])
plt.show()
plt.scatter(df['Income'],df['Sales'])
plt.show()
plt.scatter(df['Sales'],df['Age'])
plt.show()
```

## Box plot for the dataset

```python
df.plot.box()
plt.boxplot(df['Income'])
plt.show()
```

## Histogram chart using matplotlib

```python
#Create histogram for numeric data
df.hist( )
#show plot
plt.show()
```

## using seaborn library for line plot 

```python
import seaborn as sns

sns.set_theme(style="whitegrid")

sns.lineplot(data=df, x="Age", y="Sales", marker="o")

plt.title("Age vs Sales")
plt.xlabel("Age")
plt.ylabel("Sales")
plt.show()
```
