## Imputation stratgies when can be used:
 * Mean : Replace the missing value with average of column,this method can be used on Extreme values, columns such as weight, Total_sales, etc.

 * Median: Replace the missing value with middle values, this method is can be used on skewed distrbutions or wiht extreme outliers column.

 * Mode: Replace the missing values with most repeated values, this method can be used on boolean columns, such as gender, yes/no question columns.

 * Forward Fill : Replaces missing values with the previous available value. It is useful when values remain constant, such as a temperature reading between updates.

```python
##Mean fill
df["Postal Code"].fillna(int(df["Postal Code"].mean()))
```

```python
##Median fill
df["Postal Code"].fillna(int(df["Postal Code"].median()))
```

```python
##Mean fill
df["Postal Code"].fillna(int(df["Postal Code"].mode()))
```

```python
## Forward fill
df["Postal Code"].ffill()
```

```python
## Calculating the Q1 and Q3 quartiles
Q1 = df['Sales'].quantile(0.25)
Q3 = df['Sales'].quantile(0.75)
```

```python
#Calculating Inter quartile range
IQR = Q3 - Q1

# setting upper and lower bound
lower_bound = Q1 - (1.5 * IQR)
upper_bound = Q3 + (1.5 * IQR)
```

```python
# Removing outliers using upper and lower bound
df_clean = df[
    (df['Sales'] >= lower_bound) & (df['Sales'] <= upper_bound)
]
```

```python
# Create an empty list for the data quality log
quality_log = []

#Check dataset size
quality_log.append({
    "Check": "Dataset Size",
    "Finding": f"{df.shape[0]} rows and {df.shape[1]} columns",
    "Action": "Recorded dataset dimensions"
})

#Check missing values
for col in df.columns:
    missing = df[col].isnull().sum()

    if missing > 0:
        quality_log.append({
            "Check": "Missing Values",
            "Finding": f"{col}: {missing} missing values",
            "Action": "Investigate and handle missing values"
        })

#Check duplicate rows
duplicates = df.duplicated().sum()

quality_log.append({
    "Check": "Duplicate Rows",
    "Finding": f"{duplicates} duplicate rows",
    "Action": (
        "Remove duplicates if inappropriate"
        if duplicates > 0
        else "No action needed"
    )
})

#Check data types
for col in df.columns:
    quality_log.append({
        "Check": "Data Type",
        "Finding": f"{col}: {df[col].dtype}",
        "Action": "Verify the expected data type"
    })

#Check string inconsistencies
text_columns = df.select_dtypes(
    include=["object", "string"]
).columns

for col in text_columns:
    values = df[col].dropna().astype(str)

    whitespace = (values != values.str.strip()).sum()
    inconsistent_case = (
        values != values.str.lower()
    ).sum()

    if whitespace > 0:
        quality_log.append({
            "Check": "Whitespace",
            "Finding": f"{col}: {whitespace} values have extra surrounding spaces",
            "Action": "Strip leading and trailing spaces"
        })

    if inconsistent_case > 0:
        quality_log.append({
            "Check": "Casing",
            "Finding": f"{col}: {inconsistent_case} values are not lowercase",
            "Action": "Review and standardize casing if appropriate"
        })

#Check outliers using the IQR method
numeric_columns = df.select_dtypes(
    include="number"
).columns

for col in numeric_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = (
        (df[col] < lower_bound) |
        (df[col] > upper_bound)
    ).sum()

    quality_log.append({
        "Check": "Outliers (IQR)",
        "Finding": f"{col}: {outliers} potential outliers",
        "Action": "Investigate before removing"
    })

#Convert the log into a DataFrame
log_df = pd.DataFrame(quality_log)

#Export the log to Excel
log_df.to_excel(
    "data_quality_log.xlsx",
    index=False
)

print(log_df)
print("Data quality log exported successfully!")
```
