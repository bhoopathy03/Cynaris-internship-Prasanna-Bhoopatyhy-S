```python
n=2000

df = pd.DataFrame({
	"age": np.random.randint(20,70,size=n),
	"income": np.random.normal(60000,15000,size=n).round(2),
	"loan_amount":np.random.normal(15000,5000,size=n).round(2),
	"credit score":np.random.randint(300,850,size=n),
	"employment_status": np.random.choice(["Employed","unemployed","self-employed"],size=n),
	"loan_approved":np.random.choice([0,1],size=n,p=[0.4,0.6])
})
```

```python
report=sv.analyze(df)
report.show_html("report.html")
```

```python
from autoviz import AutoViz_Class

AV = AutoViz_Class()
```

```python
dft = AV.AutoViz(
    "",
    dfte=df,
    depVar="loan_approved",
    verbose=1,
    chart_format="html"
)
```

```python
import os
import glob
import webbrowser
import pandas as pd
from autoviz import AutoViz_Class

AV = AutoViz_Class()

dft = AV.AutoViz(
    "",
    dfte=df,
    depVar="loan_approved",
    verbose=1,
    chart_format="html"
)

# Find the generated HTML file
html_files = glob.glob("AutoViz/**/*.html", recursive=True)

# Open the HTML report
if html_files:
    html_path = os.path.abspath(html_files[0])
    webbrowser.open("file://" + html_path)
    print("HTML report opened:", html_path)
else:
    print("No HTML file found. Check the AutoViz output folder.")
```
